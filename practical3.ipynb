```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def tanh(x):
    return np.tanh(x)
```

```python
def relu(x):
    return np.maximum(0,x)
```

```python
def softmax(x):
    exp_x=np.exp(x-np.max(x))
    return exp_x/np.sum(exp_x)
```
