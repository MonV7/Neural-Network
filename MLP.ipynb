```python
import numpy as np
```

```python
x=np.array([[0,0],[0,1],[1,0],[1,1]],dtype=float)

Y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)
```

```python
w1=np.random.randn(2,4)*0.5
b1=np.zeros((1,4))

w2=np.random.randn(4,1)*0.5
b2=np.zeros((1,1))
```

```python

# sigmoid function 
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x*(1-x)
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
# Training
learning_rate = 0.1
epochs = 10000
```

```python
for epoch in range(epochs):

    # Forward propagation
    Z1 = X @ W1 + b1

    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2

    Y_pred = sigmoid(Z2)

    # Binary cross-entropy loss
    epsilon = 1e-8

    loss = -np.mean(
        Y * np.log(Y_pred + epsilon)
        + (1 - Y) * np.log(1 - Y_pred + epsilon)
    )

    # Backpropagation
    dZ2 = (Y_pred - Y) / len(X)

    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = dZ2 @ W2.T

    dZ1 = dA1 * relu_derivative(Z1)

    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    # Update weights
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1
```

```python
# Final prediction
Z1 = X @ W1 + b1
A1 = relu(Z1)

Z2 = A1 @ W2 + b2
Y_pred = sigmoid(Z2)

print("Predictions:")
print(Y_pred)
```
